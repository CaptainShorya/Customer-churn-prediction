```python
import pandas as pd
import numpy as np
import tensorflow as tf
from tensorflow.keras.models import load_model
import pickle
```

```python
## Load the model, scaler and encoder
model = load_model('model.h5')

with open('label_encoder.pkl','rb') as file:
    label_encoder=pickle.load(file)

with open('ohe_geography.pkl','rb') as file:
    ohe_geography=pickle.load(file)

with open('StandardScaler.pkl','rb') as file:
    standardScaler=pickle.load(file)  
```

```python
## Input data
input_data = {
    'CreditScore': 600,
    'Geography': 'France',
    'Gender': 'Male',
    'Age': 40,
    'Tenure': 3,
    'Balance': 60000,
    'NumOfProducts': 2,
    'HasCrCard': 1,
    'IsActiveMember': 1,
    'EstimatedSalary': 50000
}
```

```python
## OneHotEncoder expects a 2D input
geo_encoded = ohe_geography.transform([[input_data['Geography']]])
geo_encoded_df = pd.DataFrame(geo_encoded,columns=ohe_geography.get_feature_names_out(['Geography']))
geo_encoded_df
```

```python
# Converting dictionary into dataframe
input_df = pd.DataFrame([input_data])
input_df
```

```python
# Concatenate ohe output with input_df
input_df = pd.concat([geo_encoded_df,input_df.drop(['Geography'],axis=1)],axis=1)
input_df
```

```python
# Encode Categorical features
input_df['Gender'] = label_encoder.transform(input_df['Gender'])
input_df
```

```python
# Scaling the input data
input_scaled = standardScaler.transform(input_df)
input_scaled
```

```python
## Predict the output 
output = model.predict(input_scaled)
output[0][0]
```

```python
if output[0][0] > 0.5:
    print("Prediction is likely to churn")
else:
    print("Prediction is not likely to churn")
```
